### MLOps Steps

1. Define the Problem 

> The output of the machine learning model 

> The type of machine learning task that is being implemented

> The Criteria that makes the model successful

2. Training the model 

> The type of model that needs to be trained 

> The amount of time to be invested in training the model 

> Whether full control is needed in training the model 

> Services available in the current oraganisation

> 

4. Compute Options

> <font color = "orange"> <b> CPU (Central Processing Unit) & GPU (Graphic Processing Units) :</b> </font> For smaller tabular data sets CPU is sufficient and cost effective.

> For Larger Tabular Datasets or data stored in text or images , GPUs is the way to go as it is more powerful and efficient

> <font color = "orange"> <b> General purpose or Memory Optimized :</b></font> General Purpose has a balanced ratio with CPU to Memory , this is mainly for smaller datasets for testing and deployment

> Memory Optimized is for in memory analytics basically when working with notebooks or large datasets

### MLTable

> <font color = "orange"> Azure Machine Learning supports the table type MLtable </font>

> <font color = "orange"> Allows creation of the blueprint that defines how to load the data files </font>, whether in a pandas or spark 

> The structure of the MLTable is a **YAML** based file that defines the structure of the data. within that the user can specify :
    >> 1. The location of the data , whether be in the local , cloud or in the public server
    >> 2. Based on patterns that is given below as sample
    >> 3. Read Transformation - file format type in (json,delimeter,parquet,delta) etc.
    >> 4. Column type conversion & New column conversion
    >> 5. Take random samples , keep/drop columns or filter


> <font color = "orange"><b>Globbing is a method that matches files or strings based on wildcards also known as glob matching </b>

>
